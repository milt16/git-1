```
import numpy as np
import IPython.display as display
from matplotlib import pyplot as plt
import io
import base64

ys = 200 + np.random.randn(100)
x = [x for x in range(len(ys))]

fig = plt.figure(figsize=(4, 3), facecolor='w')
plt.plot(x, ys, '-')
plt.fill_between(x, ys, 195, where=(ys > 195), facecolor='g', alpha=0.6)
plt.title("Sample Visualization", fontsize=10)

data = io.BytesIO()
plt.savefig(data)
image = F"data:image/png;base64,{base64.b64encode(data.getvalue()).decode()}"
alt = "Sample Visualization"
display.display(display.Markdown(F"""![{alt}]({image})"""))
plt.close(fig)
```

```
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
import math

# Константы
MU0 = 4 * math.pi * 1e-7  # Магнитная постоянная, Гн/м (µ0 - "мю нулевое")


# Функция для создания спирали

def generate_spiral(R, h, N_vitkov, N_points):
    """
    Генерация координат спирали
    R - радиус витка (м) - насколько далеко проволока от центра
    h - шаг спирали (м) - расстояние между соседними витками
    N_vitkov - число витков - сколько раз проволока оборачивается вокруг оси
    N_points - число точек для аппроксимации - из скольких кусочков состоит спираль

    Математика: спираль задаётся параметрически:
    x = R * cos(t) - точка бежит по окружности (косинус даёт координату X)
    y = R * sin(t) - синус даёт координату Y (вместе с cos получается окружность)
    z = h * t - а Z линейно растёт, поэтому получается не плоское кольцо, а спираль
    t меняется от 0 до 2π * число витков (полный оборот = 2π радиан = 360 градусов)
    """
    # Создаём массив значений угла t от 0 до 2π * число витков
    # Например, если 3 витка, то t от 0 до ~18.85 (6π)
    t = np.linspace(0, 2*np.pi*N_vitkov, N_points)

    # Вычисляем координаты по формулам спирали
    x = R * np.cos(t)  # X меняется по косинусу - движение по окружности
    y = R * np.sin(t)  # Y меняется по синусу - движение по окружности
    z = h * t          # Z растёт пропорционально углу - подъём спирали

    # Объединяем x, y, z в один массив, где каждая строка - одна точка (x, y, z)
    # Это стандартный формат для хранения 3D координат
    return np.column_stack((x, y, z))


# Функция для расчёта поля от одного маленького отрезка провода

def biot_savart_segment(dl, r_prime, r, I):
    """
    Вклад в магнитное поле от одного отрезка по закону Био-Савара-Лапласа

    dl - вектор длины отрезка (направление тока и длина кусочка провода)
    r_prime - координаты центра отрезка (где находится этот кусочек)
    r - координаты точки наблюдения (где мы хотим узнать поле)
    I - сила тока (сколько ампер течёт по проводу)

    Закон Био-Савара-Лапласа (упрощённо):
    Маленький кусочек провода с током создаёт магнитное поле, которое:
    1. Пропорционально току I
    2. Пропорционально длине отрезка |dl|
    3. Направлено перпендикулярно и отрезку, и линии от отрезка к точке
    4. Убывает с расстоянием как 1/r² (в формуле r³ из-за математики)
    """
    # Вектор от центра отрезка до точки наблюдения
    r_vec = r - r_prime

    # Расстояние от отрезка до точки (длина вектора r_vec)
    r_mag = np.linalg.norm(r_vec)

    # Защита от деления на ноль: если точка слишком близко к проводу
    if r_mag < 1e-6:  # 1e-6 метра = 1 микрометр
        return np.zeros(3)  # возвращаем нулевое поле

    # Векторное произведение dl × r_vec - это ключевая часть формулы
    # Оно даёт направление поля: перпендикулярно и dl, и r_vec
    cross_product = np.cross(dl, r_vec)

    # Финальная формула закона Био-Савара-Лапласа
    # MU0 * I / (4π) - константы и ток
    # cross_product - определяет направление
    # / r_mag**3 - поле убывает с расстоянием
    return MU0 * I / (4 * np.pi) * cross_product / (r_mag**3)


# Функция для расчёта поля от всей спирали

def magnetic_field_spiral(R, h, N_vitkov, I, points, N_segments=500):
    """
    Расчёт магнитного поля спирали в заданных точках

    R, h, N_vitkov - параметры спирали (см. generate_spiral)
    I - сила тока
    points - массив точек, в которых нужно посчитать поле (каждая строка - точка x,y,z)
    N_segments - на сколько отрезков разбить спираль (чем больше, тем точнее, но медленнее)

    Принцип: разбиваем спираль на маленькие отрезки,
    для каждого отрезка считаем поле в каждой точке наблюдения,
    суммируем все вклады (принцип суперпозиции)
    """
    # Сначала создаём спираль из N_segments точек
    spiral = generate_spiral(R, h, N_vitkov, N_segments)

    # Создаём массив для результатов такой же формы, как points
    # Изначально заполняем нулями
    B_total = np.zeros_like(points)

    # Проходим по всем отрезкам спирали
    # Отрезков будет на 1 меньше, чем точек (последняя точка не образует отрезок)
    for i in range(len(spiral)-1):
        # Вектор текущего отрезка (от точки i до точки i+1)
        dl = spiral[i+1] - spiral[i]

        # Центр отрезка (среднее между началом и концом)
        r_prime = (spiral[i] + spiral[i+1]) / 2

        # Для каждой точки наблюдения считаем вклад от этого отрезка
        for j, point in enumerate(points):
            # Добавляем вклад к общему полю в этой точке
            B_total[j] += biot_savart_segment(dl, r_prime, point, I)

    return B_total


# Главная функция для визуализации

def plot_magnetic_field(R, h, N_vitkov, I, grid_size=15):
    """
    Построение 3D визуализации магнитного поля спирали

    Параметры по умолчанию:
    R = 0.02 м = 2 см (радиус как у небольшой пружинки)
    h = 0.005 м = 0.5 см (шаг - полсантиметра между витками)
    N_vitkov = 3 (три витка)
    I = 5 А (ток 5 ампер - как в мощном блоке питания)
    grid_size = 15 (сетка 15×15×15 точек для визуализации)
    """

    # 1. Подготовка данных: создаём сетку точек для расчёта поля

    # Создаём равномерные координаты по каждой оси
    x = np.linspace(-0.05, 0.05, grid_size)  # от -5 см до +5 см по X
    y = np.linspace(-0.05, 0.05, grid_size)  # от -5 см до +5 см по Y
    z = np.linspace(-0.02, 0.02, grid_size)  # от -2 см до +2 см по Z (спираль выше, но нас интересует центр)

    # Создаём 3D сетку координат (три массива, задающие все точки куба)
    X, Y, Z = np.meshgrid(x, y, z)

    # Преобразуем сетку в список точек: каждая строка - (x, y, z) одной точки
    # Всего будет grid_size³ точек (например, 15³ = 3375 точек)
    points = np.column_stack((X.ravel(), Y.ravel(), Z.ravel()))

    print("Расчет магнитного поля...")
    print(f"  Сетка: {grid_size}×{grid_size}×{grid_size} = {len(points)} точек")
    print(f"  Параметры спирали: R={R*100:.1f} см, h={h*100:.2f} см, витков={N_vitkov}, I={I} А")

    # Рассчитываем магнитное поле во всех точках сетки
    # Результат: для каждой точки получаем вектор (Bx, By, Bz)
    B_vectors = magnetic_field_spiral(R, h, N_vitkov, I, points)

    # Вычисляем модуль (длину) вектора для каждой точки - это величина поля
    # np.linalg.norm с axis=1 вычисляет длину каждого вектора в массиве
    B_magnitude = np.linalg.norm(B_vectors, axis=1)


    # 2. Создание фигуры с четырьмя подграфиками

    fig = plt.figure(figsize=(15, 12))  # размер фигуры 15×12 дюймов


    # 3. Первый подграфик: геометрия спирали (3D вид)

    ax1 = fig.add_subplot(221, projection='3d')  # 2x2 сетка, первая позиция, 3D оси

    # Создаём детальную спираль для красивого отображения
    t_plot = np.linspace(0, 2*np.pi*N_vitkov, 200)  # 200 точек для плавной линии
    x_spiral = R * np.cos(t_plot)
    y_spiral = R * np.sin(t_plot)
    z_spiral = h * t_plot

    # Рисуем спираль красной линией толщиной 3 пикселя
    ax1.plot(x_spiral, y_spiral, z_spiral, 'r-', linewidth=3, label=f'Спираль (I={I}A)')

    # Настраиваем внешний вид
    ax1.set_title('Геометрия спирали')
    ax1.set_xlabel('X, м')
    ax1.set_ylabel('Y, м')
    ax1.set_zlabel('Z, м')
    ax1.legend()


    # 4. Второй подграфик: карта поля в плоскости XY (вид сверху)

    ax2 = fig.add_subplot(222)

    # Извлекаем срез данных при z=0 (середина по Z)
    # reshape превращает одномерный массив обратно в 3D сетку
    # grid_size//2 - индекс центрального слоя по Z (где z=0)
    Z_slice = B_magnitude.reshape(grid_size, grid_size, grid_size)[:, :, grid_size//2]

    # Строим цветную контурную карту
    # contourf - заливка цветом, levels=20 - 20 уровней яркости
    # .T нужно для правильной ориентации (транспонирование)
    im2 = ax2.contourf(x, y, Z_slice.T, levels=20, cmap='hot')  # hot - от чёрного к жёлтому

    # Добавляем белые контурные линии поверх для лучшей читаемости
    ax2.contour(x, y, Z_slice.T, levels=10, colors='white', alpha=0.5)

    # Добавляем цветовую шкалу
    plt.colorbar(im2, ax=ax2, label='Магнитная индукция, Тл (Тесла)')

    ax2.set_title('Распределение |B| в плоскости XY (z=0)')
    ax2.set_xlabel('X, м')
    ax2.set_ylabel('Y, м')

    # 5. Третий подграфик: векторное поле в сечении XZ
    #    Показываем направление магнитных силовых линий

    ax3 = fig.add_subplot(223)

    # Для наглядности берём каждый второй индекс по X и Z
    x_indices = np.arange(0, grid_size, 2)  # индексы для X: 0,2,4,...
    z_indices = np.arange(0, grid_size, 2)  # индексы для Z: 0,2,4,...
    y_idx = grid_size // 2                   # индекс для Y=0 (середина)

    # Перестраиваем массив векторов в 4D форму: (grid_size, grid_size, grid_size, 3)
    # Теперь B_reshaped[i,j,k] - вектор поля в точке (x[i], y[j], z[k])
    B_reshaped = B_vectors.reshape(grid_size, grid_size, grid_size, 3)

    # Используем np.ix_ для выбора нужных индексов
    # Выбираем: все z из z_indices, y только y_idx, все x из x_indices
    B_slice = B_reshaped[np.ix_(z_indices, [y_idx], x_indices)]  # форма: (len(z_indices), 1, len(x_indices), 3)

    # Извлекаем компоненты Bx (индекс 0) и Bz (индекс 2)
    # Убираем измерение с размером 1 (y)
    Bx_slice = B_slice[:, 0, :, 0]  # Bx для каждой точки (z,x)
    Bz_slice = B_slice[:, 0, :, 2]  # Bz для каждой точки (z,x)

    # Создаём сетку координат для отображения стрелок
    X_slice, Z_slice = np.meshgrid(x[x_indices], z[z_indices])

    # Рисуем стрелки (векторы) функцией quiver
    # stride=2 дополнительно прореживает для наглядности
    stride = 2
    ax3.quiver(X_slice[::stride, ::stride], Z_slice[::stride, ::stride],
               Bx_slice[::stride, ::stride], Bz_slice[::stride, ::stride],
               scale=1e-5, color='blue')  # scale регулирует длину стрелок

    ax3.set_title('Векторное поле B в сечении XZ (стрелки показывают направление)')
    ax3.set_xlabel('X, м')
    ax3.set_ylabel('Z, м')


    # 6. Четвёртый подграфик: поле вдоль оси спирали

    ax4 = fig.add_subplot(224)

    # Создаём 100 точек вдоль оси Z (x=0, y=0)
    z_axis = np.linspace(-0.02, 0.02, 100)
    points_axis = np.column_stack((np.zeros(100), np.zeros(100), z_axis))

    # Считаем поле в этих точках
    B_axis = magnetic_field_spiral(R, h, N_vitkov, I, points_axis, 200)

    # Переводим в микроТесла для удобства (1 Тл = 1 000 000 мкТл)
    # Поле Земли ~50 мкТл, для сравнения
    Bz_axis = np.linalg.norm(B_axis, axis=1) * 1e6

    # Рисуем график: Z в сантиметрах, |B| в мкТл
    ax4.plot(z_axis*100, Bz_axis, 'g-', linewidth=2, label='|B| вдоль оси Z')
    ax4.set_xlabel('Z, см')
    ax4.set_ylabel('|B|, мкТл')
    ax4.set_title('Магнитное поле вдоль оси спирали')
    ax4.grid(True)  # добавляем сетку
    ax4.legend()


    # 7. Настройка и сохранение

    plt.tight_layout()  # автоматически подгоняем расположение графиков

    # Сохраняем в файл с высоким разрешением
    plt.savefig('magnetic_field_spiral.png', dpi=300, bbox_inches='tight')
    print("График сохранён в файл 'magnetic_field_spiral.png'")

    # Показываем окно с графиками
    plt.show()

    # Выводим численные результаты в консоль
    print(f"Максимальная магнитная индукция в исследуемой области: {np.max(B_magnitude)*1e6:.1f} мкТл")
    print(f"Индукция в центре (z=0): {Bz_axis[len(Bz_axis)//2]:.1f} мкТл")
    print(f"Для сравнения: магнитное поле Земли ~ 50 мкТл")


# Запуск программы
# Задаём параметры спирали (в метрах)
R = float(input('Введите радиус (в метрах (пример 0.02))'))       # радиус 2 см (примерно как пальчиковая батарейка)
h = float(input('Введите шаг (в метрах)'))      # шаг 0.5 см (полсантиметра между витками)
N_vitkov = int(input('Введите количество витков'))   # 3 витка
I = float(input('Введите значение тока'))          # ток 5 А (как в заряднике для ноутбука)

# Вызываем главную функцию
plot_magnetic_field(R, h, N_vitkov, I)
```
